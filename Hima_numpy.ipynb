## A3. For compatible 2D arrays A and B, explain A * B vs A @ B. Describe one neural-network situation for each.

`A * B` performs element-wise multiplication, so corresponding elements are multiplied. `A @ B` performs matrix multiplication, where rows and columns are multiplied according to matrix multiplication rules. Element-wise multiplication can be used for applying a mask or scaling values. Matrix multiplication is commonly used when calculating the output of a neural-network layer using inputs and weights.

```python
#Example Program

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[2, 2],
              [2, 2]])

print(A * B)
print(A @ B)

#Example explanation: * multiplies corresponding elements, while @ performs matrix multiplication.
```

## A4. What is broadcasting in NumPy? State the general rule used to decide whether arrays of different shapes can be combined element-wise.

Broadcasting allows NumPy to perform element-wise operations between arrays having different but compatible shapes. NumPy compares their dimensions starting from the last dimension. Two dimensions are compatible when they are equal or when one of them is 1. This allows smaller arrays to work with larger arrays without manually copying the values.

```python
#Example Program

marks = np.array([10, 20, 30])

result = marks + 5

print(result)

#Example explanation: The value 5 is automatically applied to every element of the array.
```

## A5. Explain flatten() vs ravel(). In a memory-constrained application processing large images, which would you prefer and why?

Both `flatten()` and `ravel()` convert an array into a one-dimensional array. `flatten()` normally creates a new copy of the data, while `ravel()` tries to return a view when possible. Therefore, `ravel()` generally uses less extra memory. For large images with limited memory, I would prefer `ravel()`.

```python
#Example Program

arr = np.array([[1, 2],
                [3, 4]])

print(arr.flatten())
print(arr.ravel())

#Example explanation: Both convert the array into one dimension, but ravel() can avoid creating an extra copy when possible.
```

## A6. For a 2D array, explain axis=0 and axis=1 in an aggregation function such as np.sum() or np.mean().

For a 2D array, `axis=0` means the operation is performed down the rows, giving a result for each column. `axis=1` means the operation is performed across the columns, giving a result for each row.

For example:

arr = np.array([[1, 2, 3],
                [4, 5, 6]])

np.sum(arr, axis=0) gives [5, 7, 9], while np.sum(arr, axis=1) gives [6, 15].

```python
#Example Program

arr = np.array([[1, 2, 3],
                [4, 5, 6]])

print(np.sum(arr, axis=0))
print(np.sum(arr, axis=1))

#Example explanation: axis=0 gives the sum of each column and axis=1 gives the sum of each row.
```

## A7. Why is it important to normalize or standardize numerical features before training most ML models? Explain normalization vs standardization.

Normalization or standardization makes numerical features have a suitable scale before giving them to a model. This prevents features with large values from having too much influence compared to smaller ones. Normalization usually changes values to a fixed range such as 0 to 1. Standardization changes values based on their mean and standard deviation.

```python
#Example Program

marks = np.array([20, 40, 60, 80, 100])

normalized = (marks - marks.min()) / (marks.max() - marks.min())

standardized = (marks - marks.mean()) / marks.std()

print(normalized)
print(standardized)

#Example explanation: Normalization changes the values to a 0–1 range, while standardization changes them based on the mean and standard deviation.
```

## A8. Why do functions such as np.nanmean() exist? What happens if np.mean() is used directly on data containing NaN?

`np.nanmean()` is used to calculate the mean while ignoring `NaN` values. If `np.mean()` is directly used on an array containing `NaN`, the result will normally also be `NaN`. Therefore, `np.nanmean()` is useful when some data values are missing.

```python
#Example Program

data = np.array([10, 20, np.nan, 40])

print(np.mean(data))
print(np.nanmean(data))

#Example explanation: np.mean() is affected by NaN, while np.nanmean() ignores the missing value.
```

## D1. Data Preprocessing Pipeline

We have a NumPy array containing NaN values. The steps are:

1.Detect the missing values using np.isnan().

2.Find the mean of the available values using np.nanmean().

3.Replace the missing values with the mean.

4.Standardize the cleaned data using mean and standard deviation.

```python
data = np.array([10, 20, np.nan, 40, 50])

# Detect missing values
missing = np.isnan(data)
print("Missing values:", missing)

# Find the mean without considering NaN
mean_value = np.nanmean(data)

# Replace NaN with the mean
data[missing] = mean_value
print("Cleaned data:", data)

# Standardize the clean data
standardized = (data - np.mean(data)) / np.std(data)

print("Standardized data:", standardized)
```

### Explanation:
np.isnan() is used to find the missing values. I used np.nanmean() to calculate the mean without including NaN, and replaced the missing value with this mean. Finally, the cleaned data is standardized using its mean and standard deviation.

### Justification:
I used mean replacement because it keeps the data and does not remove the value completely. After handling the missing value, the data can be standardized before giving it to the model.

## D3. Debug the Model Code

The given code contains more than one bug.

```python
X = np.array([[1, 2],
              [2, 3],
              [3, 4]])

w = np.array([0.5, 1.0])

b = 2.0

y_true = np.array([10, 20, 30])

y_pred = X @ w + b

mse = np.mean((y_true - y_pred) ** 2)

print("Predicted values:", y_pred)
print("MSE:", mse)
```

## Bug 1 – Weight Shape

X has shape (3, 2), which means there are 3 samples and 2 features.

But w has shape (3,), so it contains 3 weights.

For X @ w, the number of features in X must match the number of values in w.

So w should contain 2 weights.

## Bug 2 – MSE Formula

The given formula is:

np.mean(y_true - y_pred ** 2)

Because of operator precedence, y_pred ** 2 is calculated first.

So it is actually calculating:

np.mean(y_true - (y_pred ** 2))

For MSE, we need to first find the difference between actual and predicted values, then square the complete difference:

np.mean((y_true - y_pred) ** 2)

```python
X = np.array([[1, 2],
              [2, 3],
              [3, 4]])

w = np.array([0.5, 1.0])

b = 2.0

y_pred = X @ w + b

y_true = np.array([10, 20, 30])

mse = np.mean((y_true - y_pred) ** 2)

print("Predicted values:", y_pred)
print("MSE:", mse)

```

### Explanation:
X has 2 features, so w must have 2 values. The MSE formula also needs parentheses so that the complete difference between y_true and y_pred is squared before calculating the mean.

### Justification:
The first bug is the mismatch between the number of features and weights. The second bug is the missing parentheses in the MSE formula, because the complete prediction error should be squared.
